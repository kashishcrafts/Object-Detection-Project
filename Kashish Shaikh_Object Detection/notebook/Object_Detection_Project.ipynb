```python
# 3
import fiftyone as fo
import fiftyone.zoo as foz

TARGET_CLASSES = ["Person", "Laptop", "Chair"]

if "open-images-v7-train-1500" in fo.list_datasets():
    fo.delete_dataset("open-images-v7-train-1500")

print("Streaming training split from Open Images V7 (1500 samples)")
train_dataset = foz.load_zoo_dataset(
    "open-images-v7",
    split="train",
    label_types="detections",
    classes=TARGET_CLASSES,
    max_samples=1500,
    dataset_name="open-images-v7-train-1500",
    drop_existing=True
)

print("\nDownloading validation split")
val_dataset = foz.load_zoo_dataset(
    "open-images-v7",
    split="validation",
    label_types="detections",
    classes=TARGET_CLASSES,
    max_samples=100,
    dataset_name="open-images-v7-validation-100",
    drop_existing=True
)

print("\nData layer fully synchronized!")
```

```python
import numpy as np
import cv2
import fiftyone as fo

MAX_LABELED_OBJECTS = 10

def load_pure_dataset(split_name):
    images_list = []
    boxes_list = []
    labels_list = []

    if split_name == "train":
        fo_dataset = fo.load_dataset("open-images-v7-train-1500")
    else:
        fo_dataset = fo.load_dataset("open-images-v7-validation-100")

    class_map = {"Person": 0, "Laptop": 1, "Chair": 2}
    print(f"Parsing actual Kaggle annotations for {split_name} split...")

    for sample in fo_dataset:
        img_path = sample.filepath
        img = cv2.imread(img_path)
        if img is None: continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (320, 320)) / 255.0

        real_boxes = np.zeros((MAX_LABELED_OBJECTS, 4), dtype=np.float32)
        real_labels = np.ones((MAX_LABELED_OBJECTS,), dtype=np.int64) * -1

        if sample.ground_truth is not None:
            count = 0
            for detection in sample.ground_truth.detections:
                if count >= MAX_LABELED_OBJECTS: break
                if detection.label in class_map:
                    xmin, ymin, w, h = detection.bounding_box
                    real_boxes[count] = [ymin, xmin, ymin + h, xmin + w]
                    real_labels[count] = class_map[detection.label]
                    count += 1

        images_list.append(img)
        boxes_list.append(real_boxes)
        labels_list.append(real_labels)

    return np.array(images_list, dtype=np.float32), np.array(boxes_list, dtype=np.float32), np.array(labels_list, dtype=np.int64)

X_train, Y_boxes_train, Y_labels_train = load_pure_dataset("train")
X_val, Y_boxes_val, Y_labels_val = load_pure_dataset("validation")

Y_labels_train[Y_labels_train == -1] = 3
Y_labels_val[Y_labels_val == -1] = 3

print(f"\nData parsed successfully!")
print(f"Train Images Shape: {X_train.shape} | Train Boxes Shape: {Y_boxes_train.shape}")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_accurate_detector():
    base_model = tf.keras.applications.MobileNetV2(input_shape=(320, 320, 3), include_top=False, weights='imagenet')

    base_model.trainable = True

    fine_tune_at = len(base_model.layers) - 40
    for layer in base_model.layers[:fine_tune_at]:
        layer.trainable = False

    x = base_model.output
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(512, activation='relu')(x)
    x = layers.Dropout(0.2)(x)

    box_output = layers.Dense(10 * 4, activation='sigmoid', name='boxes_head')(x)
    box_output = layers.Reshape((10, 4), name='boxes')(box_output)

    class_output = layers.Dense(10 * 4, activation='linear', name='labels_head')(x)
    class_output = layers.Reshape((10, 4), name='labels')(class_output)

    return models.Model(inputs=base_model.input, outputs={'boxes': box_output, 'labels': class_output})

def masked_box_loss(y_true, y_pred):
    raw_loss = tf.square(y_true - y_pred)
    coordinate_sum = tf.reduce_sum(tf.abs(y_true), axis=-1)
    mask = tf.cast(coordinate_sum > 0.0, tf.float32)
    mask = tf.expand_dims(mask, axis=-1)
    return tf.reduce_mean(raw_loss * mask)

custom_model = build_accurate_detector()
custom_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=5e-6),
    loss={
        'boxes': masked_box_loss,
        'labels': tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
    },
    loss_weights={'boxes': 20.0, 'labels': 1.0}
)
print("Synchronized and compiled successfully!")
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

callbacks_list = [
    EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7, verbose=1)
]

print("Launching synchronized fine-tuning pass...")
history = custom_model.fit(
    X_train,
    {'boxes': Y_boxes_train, 'labels': Y_labels_train},
    validation_data=(X_val, {'boxes': Y_boxes_val, 'labels': Y_labels_val}),
    epochs=50,
    batch_size=8,
    callbacks=callbacks_list
)
print("\nModel fully trained and best weights restored!")
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

CUSTOM_LABELS = {0: "Person", 1: "Laptop", 2: "Chair", 3: "Background"}

def verify_image(image_path, confidence_floor=0.30, nms_overlap_limit=0.3):
    img_bgr = cv2.imread(image_path)
    if img_bgr is None:
        print(f"Error: Could not locate file {image_path}")
        return

    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    h_orig, w_orig, _ = img_rgb.shape

    img_resized = cv2.resize(img_rgb, (320, 320)) / 255.0
    input_tensor = np.expand_dims(img_resized, axis=0)

    predictions = custom_model.predict(input_tensor, verbose=0)
    pred_boxes = predictions['boxes'][0]
    pred_labels = predictions['labels'][0]

    boxes, confidences, class_ids = [], [], []

    for i in range(len(pred_boxes)):
        logits = pred_labels[i]
        exp_logits = np.exp(logits - np.max(logits))
        probs = exp_logits / np.sum(exp_logits)

        class_id = int(np.argmax(probs[:3]))
        confidence = float(probs[class_id])

        if confidence > confidence_floor:
            ymin, xmin, ymax, xmax = pred_boxes[i]

            x1 = int(xmin * w_orig)
            y1 = int(ymin * h_orig)
            w = int((xmax - xmin) * w_orig)
            h = int((ymax - ymin) * h_orig)

            if w > 20 and h > 20:
                boxes.append([x1, y1, w, h])
                confidences.append(confidence)
                class_ids.append(class_id)

    indices = cv2.dnn.NMSBoxes(boxes, confidences, confidence_floor, nms_overlap_limit)

    detected_count = 0
    if len(indices) > 0:
        for i in indices.flatten():
            x, y, w, h = boxes[i]
            detected_count += 1

            color_map = {0: (0, 255, 0), 1: (255, 0, 0), 2: (0, 0, 255)}
            box_color = color_map.get(class_ids[i], (255, 255, 255))

            label_text = f"{CUSTOM_LABELS[class_ids[i]]}: {confidences[i]*100:.1f}%"
            cv2.rectangle(img_bgr, (x, y), (x + w, y + h), box_color, 3)
            cv2.putText(img_bgr, label_text, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, box_color, 2)

    output_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 10))
    plt.imshow(output_rgb)
    plt.axis('off')
    plt.show()
    print(f"Done! Cleaned up and visualized {detected_count} custom localized objects.")

verify_image("test1.jpg", confidence_floor=0.30, nms_overlap_limit=0.3)
```
